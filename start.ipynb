```python
# ============================================================
# FORCE DELETE — Navigates to HOME first, then deletes
# ============================================================
import shutil
import os

# Step 1: Go to home directory (escape any subfolder)
home = os.path.expanduser("~")
os.chdir(home)
print(f"📍 Moved to home: {os.getcwd()}")

# Step 2: List EVERYTHING to find the exact folder name
print("\n📂 All files/folders in home:")
for item in sorted(os.listdir(".")):
    print(f"   {'📁' if os.path.isdir(item) else '📄'} {item}")

# Step 3: Force delete ALL possible repo folder names
TARGETS = [
    "Ekegusii-LLM-Translation-main",
    "Ekegusii-LLM-Translation",
    "repo.zip",
]
for target in TARGETS:
    if os.path.isdir(target):
        shutil.rmtree(target, ignore_errors=True)
        print(f"\n✅ DELETED folder: {target}/")
    elif os.path.isfile(target):
        os.remove(target)
        print(f"\n✅ DELETED file: {target}")

# Step 4: Also scan for ANY folder starting with "Ekegusii"
for item in os.listdir("."):
    if item.lower().startswith("ekegusii") and os.path.isdir(item):
        shutil.rmtree(item, ignore_errors=True)
        print(f"✅ DELETED: {item}/")

print("\n📂 Files remaining:")
for item in sorted(os.listdir(".")):
    print(f"   {'📁' if os.path.isdir(item) else '📄'} {item}")

print("\n✅ DONE! Home is clean.")
```

```python
# ============================================================
# FRESH CLONE & ENVIRONMENT SETUP
# ============================================================
import urllib.request, zipfile, os, sys, site, glob

# Step 1: Purge any cached 'src' modules from memory
for mod in list(sys.modules.keys()):
    if mod.startswith('src'):
        del sys.modules[mod]

# Step 2: Go to home directory
home = os.path.expanduser('~')
os.chdir(home)
print(f'📍 Starting from home: {os.getcwd()}')

# Step 3: Download fresh repo ZIP from GitHub
print('📥 Downloading latest repository code from GitHub main branch...')
zip_path = os.path.join(home, 'repo.zip')
urllib.request.urlretrieve(
    'https://github.com/aykahsay/Ekegusii-LLM-Translation/archive/refs/heads/main.zip',
    zip_path
)
print('✅ Download complete!')

# Step 4: Extract and enter project directory
print('📦 Extracting...')
with zipfile.ZipFile(zip_path, 'r') as z:
    z.extractall(home)
os.remove(zip_path)
print('✅ Extracted!')

proj_dir = os.path.join(home, 'Ekegusii-LLM-Translation-main')
os.chdir(proj_dir)
if proj_dir not in sys.path:
    sys.path.insert(0, proj_dir)

# Step 5: Configure sys.path for user and conda site-packages
user_site = site.getusersitepackages()
if user_site not in sys.path:
    sys.path.insert(0, user_site)
for conda_site in glob.glob('/opt/conda/lib/python3.*/site-packages'):
    if conda_site not in sys.path:
        sys.path.insert(0, conda_site)

print(f'\n📁 Working Directory : {os.getcwd()}')
print(f'🐍 Python Executable  : {sys.executable}')
print('\n✅ FRESH CLONE & SETUP COMPLETE!')
print('🚀 You can now open any notebook in notebooks/ (e.g. notebooks/00_setup_environment.ipynb)!')
```
